<h2>Classification</h2><br>
Supervised learning technique used to predict classes (discrete data/categorical data). <br>
Balancing the dateset: Before we train the model, make sure all the classes have equal say (have equal number of rows of data) in the model building. 


```python
#Libraries
from sklearn.datasets import load_breast_cancer
import pandas as pd

from sklearn.model_selection import train_test_split
#  !pip install imbalanced-learn
from imblearn.over_sampling import RandomOverSampler, SMOTE
from imblearn.under_sampling import RandomUnderSampler

from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier

#Evaluation Metric
from sklearn.metrics import confusion_matrix, accuracy_score, precision_score, recall_score, f1_score, roc_auc_score, classification_report

```

```python
#load the dataset
data = load_breast_cancer()
df = pd.DataFrame(data.data, columns=data.feature_names)
df['target'] = data.target
print(df.head(3))
print("Columns: ",df.columns)
print("\n Target Classes: \n",data.target_names)
print("\n Target classes value counts:\n",df['target'].value_counts())
print("\n Percentage Distribution:\n", df['target'].value_counts(normalize=True)*100)

print("This is Imbalanced Dataset, since the % distribution is different.")
```

```python
X = df.drop("target",axis=1)
y = df["target"]
X_train, X_test, y_train, y_test = train_test_split(X,y,test_size=0.2, random_state=10)
print("\n Target classes value counts:\n",y_train.value_counts())
print("\n Percentage Distribution:\n", y_train.value_counts(normalize=True)*100)
```

```python
# Method 1: OverSampling using RandomOverSampler
ros = RandomOverSampler()
X_train_balanced, y_train_balanced = ros.fit_resample(X_train, y_train)
print("\n Target classes value counts:\n",y_train_balanced.value_counts())
print("\n Percentage Distribution:\n", y_train_balanced.value_counts(normalize=True)*100)
```

```python
# Method 2: UnderSampling using RandomUnderSampler
rus = RandomUnderSampler()
X_train_balanced, y_train_balanced = rus.fit_resample(X_train, y_train)
print("\n Target classes value counts:\n",y_train_balanced.value_counts())
print("\n Percentage Distribution:\n", y_train_balanced.value_counts(normalize=True)*100)
```

```python
# Method 3: OverSampling using SMOTE
smote = SMOTE()
X_train_balanced, y_train_balanced = smote.fit_resample(X_train, y_train)
print("\n Target classes value counts:\n",y_train_balanced.value_counts())
print("\n Percentage Distribution:\n", y_train_balanced.value_counts(normalize=True)*100)
```

```python
# Now the dataset is balanced, we can move ahead with model building
#Preprocessing - scaling
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train_balanced)
X_test_scaled = scaler.fit_transform(X_test)
```

```python
# MODEL 1: LOGISTIC REGRESSION
model = LogisticRegression()
model.fit(X_train_scaled, y_train_balanced)  #train the model
y_pred_train = model.predict(X_train_scaled)
print("Training data accuracy: ",accuracy_score(y_train_balanced,y_pred_train))

y_pred_test =  model.predict(X_test_scaled)
print("\n CONFUSION MATRIX - Test Data:\n",confusion_matrix(y_test,y_pred_test))
'''
TP   FP
FN   TN
'''
print("Test data accuracy: ",accuracy_score(y_test,y_pred_test))
print("Test data precision: ",precision_score(y_test,y_pred_test))
print("Test data recall: ",recall_score(y_test,y_pred_test))
print("Test data F1 Score: ",f1_score(y_test,y_pred_test))
print("Test data roc_auc_score: ",roc_auc_score(y_test,y_pred_test))
print("Test data classification report: \n",classification_report(y_test,y_pred_test))

```

```python
# MODEL 2: K NEAREST NEIGHBORS
model = KNeighborsClassifier(n_neighbors = 11)
model.fit(X_train_scaled, y_train_balanced)  #train the model
y_pred_train = model.predict(X_train_scaled)
print("Training data accuracy: ",accuracy_score(y_train_balanced,y_pred_train))

y_pred_test =  model.predict(X_test_scaled)
print("\n CONFUSION MATRIX - Test Data:\n",confusion_matrix(y_test,y_pred_test))
'''
TP   FP
FN   TN
'''
print("Test data accuracy: ",accuracy_score(y_test,y_pred_test))
print("Test data precision: ",precision_score(y_test,y_pred_test))
print("Test data recall: ",recall_score(y_test,y_pred_test))
print("Test data F1 Score: ",f1_score(y_test,y_pred_test))
print("Test data roc_auc_score: ",roc_auc_score(y_test,y_pred_test))
print("Test data classification report: \n",classification_report(y_test,y_pred_test))

#Fine tuning KNN method
for i in range(1,20,2):
    model = KNeighborsClassifier(n_neighbors = i)
    model.fit(X_train_scaled, y_train_balanced)  #train the model
    y_pred_test =  model.predict(X_test_scaled)
    print("Test data accuracy for",i,": ",accuracy_score(y_test,y_pred_test))
```

```python
# MODEL 3: DECISION TREE CLASSIFIER
model = DecisionTreeClassifier()
model.fit(X_train_scaled, y_train_balanced)  #train the model
y_pred_train = model.predict(X_train_scaled)
print("Training data accuracy: ",accuracy_score(y_train_balanced,y_pred_train))

y_pred_test =  model.predict(X_test_scaled)
print("\n CONFUSION MATRIX - Test Data:\n",confusion_matrix(y_test,y_pred_test))
'''
TP   FP
FN   TN
'''
print("Test data accuracy: ",accuracy_score(y_test,y_pred_test))
print("Test data precision: ",precision_score(y_test,y_pred_test))
print("Test data recall: ",recall_score(y_test,y_pred_test))
print("Test data F1 Score: ",f1_score(y_test,y_pred_test))
print("Test data roc_auc_score: ",roc_auc_score(y_test,y_pred_test))
print("Test data classification report: \n",classification_report(y_test,y_pred_test))
```

```python
# MODEL 4: RANDOM TREE CLASSIFIER
model = RandomForestClassifier(n_estimators = 501)
model.fit(X_train_scaled, y_train_balanced)  #train the model
y_pred_train = model.predict(X_train_scaled)
print("Training data accuracy: ",accuracy_score(y_train_balanced,y_pred_train))

y_pred_test =  model.predict(X_test_scaled)
print("\n CONFUSION MATRIX - Test Data:\n",confusion_matrix(y_test,y_pred_test))
'''
TP   FP
FN   TN
'''
print("Test data accuracy: ",accuracy_score(y_test,y_pred_test))
print("Test data precision: ",precision_score(y_test,y_pred_test))
print("Test data recall: ",recall_score(y_test,y_pred_test))
print("Test data F1 Score: ",f1_score(y_test,y_pred_test))
print("Test data roc_auc_score: ",roc_auc_score(y_test,y_pred_test))
print("Test data classification report: \n",classification_report(y_test,y_pred_test))
```
